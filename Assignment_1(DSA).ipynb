Create the 3D Array

```python
import numpy as np
data = np.arange(1, 3*16*10 + 1).reshape(3, 16, 10)
print("Shape:", data.shape)
print(data)
```

Year-wise Average

```python
year_avg = data.mean(axis=(1, 2))
print("Year-wise Average:\n", year_avg)
```

Student-wise Total

```python
student_total = data.sum(axis=2)
print("Student-wise Total:\n", student_total)
```

Subject-wise Highest Score

```python
subject_max = data.max(axis=(0, 1))
print("Subject-wise Highest Score:\n", subject_max)
```

Replace Marks Below 40 with 40

```python
data[data < 40] = 40
print("Updated Data:\n", data)
```

Sort Marks Subject-wise

```python
sorted_data = np.sort(data, axis=1)
print("Subject-wise Sorted Data:\n", sorted_data)
```
